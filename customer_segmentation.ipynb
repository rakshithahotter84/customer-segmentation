# Customer Segmentation using K-Means Clustering

This notebook explores customer data and creates customer segments using K-Means clustering.

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

## 2. Load the Dataset

```python
df = pd.read_csv('../data/Mall_Customers.csv')
df.head()
```

```python
print('Shape:', df.shape)
df.info()
print('\nMissing values:')
print(df.isnull().sum())
```

## 3. Exploratory Data Analysis

```python
df.describe()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', s=80)
plt.title('Income vs Spending Score')
plt.show()
```

## 4. Elbow Method

The Elbow Method compares K-Means inertia for different values of K. The project uses K=5 for the final segmentation.

```python
features = ['Annual Income (k$)', 'Spending Score (1-100)']
X = df[features]

inertias = []
for k in range(2, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X)
    inertias.append(model.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(2,11), inertias, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.xticks(range(2,11))
plt.show()
```

## 5. Scale the Features and Train K-Means

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)
df.head()
```

## 6. Visualize Customer Segments

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=df, x='Annual Income (k$)', y='Spending Score (1-100)', hue='Cluster', palette='deep', s=90)
plt.title('Customer Segments')
plt.show()
```

## 7. Cluster Profile

```python
profile = df.groupby('Cluster')[features].mean().round(2)
profile['Customer Count'] = df['Cluster'].value_counts().sort_index()
profile
```

## 8. Business Interpretation

Use the cluster averages to describe customer groups. For example, groups with high income and high spending can be treated as high-value customers, while high-income/low-spending groups may be candidates for targeted engagement.

```python
df.to_csv('../outputs/segmented_customers.csv', index=False)
profile.to_csv('../outputs/cluster_profile.csv')
print('Segmented data and cluster profile saved to outputs/.')
```
